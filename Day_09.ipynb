```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv("Mall_Customers.csv")
print(df.shape)
print(df.head())
print(df.describe())
print("Missing Values:", df.isnull().sum().sum())
```

```python
features = ["Annual Income (k$)", "Spending Score (1-100)"]
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


k_values = range(2, 11)
inertias = []
silhouettes = []

for k in k_values:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled, km.labels_))

fig, ax1 = plt.subplots(figsize=(10, 5))

color = 'tab:red'
ax1.set_xlabel('Number of Clusters (k)')
ax1.set_ylabel('Inertia (Elbow Method)', color=color)
ax1.plot(k_values, inertias, marker='o', color=color)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Silhouette Score', color=color)
ax2.plot(k_values, silhouettes, marker='s', color=color)
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Elbow Method and Silhouette Score for Optimal k')
fig.tight_layout()
plt.show()
```

```python
k_values = range(2, 11)
inertias = []
silhouettes = []

for k in k_values:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X_scaled)
    inertias.append(km.inertia_) # total squared distance
    silhouettes.append(silhouette_score(X_scaled, km.labels_))
    print(f"k={k:2d}  inertia={km.inertia_:10.2f}  silhouette={silhouettes[-1]:.3f}")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(list(k_values), inertias, marker='o')
ax[0].set_xlabel('Number of clusters (k)')
ax[0].set_ylabel('Inertia (WCSS)')
ax[0].set_title('Elbow method (look for the bend)')

ax[1].plot(list(k_values), silhouettes, marker='o', color='green')
ax[1].set_xlabel('Number of clusters (k)')
ax[1].set_ylabel('Silhouette score')
ax[1].set_title('Silhouette score (higher is better)')
plt.tight_layout()
plt.show()
```

```python
best_k = 5   # <- choose from the elbow / silhouette plots
kameans = KMeans(n_clusters=best_k, n_init=10, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print("Customers per cluster:\n", df['Cluster'].value_counts().sort_index())
print("Final silhouette score:", round(silhouette_score(X_scaled, df['Cluster']), 3))

# ----------------------------------------------------
# 5. VISUALIZE THE CLUSTERS
# ----------------------------------------------------
# Centroids were learned in scaled space, so convert them back to real units
centroids = scaler.inverse_transform(kmeans.cluster_centers_)

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df, x=features[0], y=features[1],
    hue="Cluster", palette="tab10", s=70
)
plt.scatter(centroids[:, 0], centroids[:, 1],
            marker="X", s=250, color="black", label="Centroids")
plt.title(f"K-means Clustering (k={best_k})")
plt.legend()
plt.show()
```
